```python
fly_num = '209'
fly_path = f'/oak/stanford/groups/trc/data/Ilana/2P/data/fly_{fly_num}'
later_dir = '/oak/stanford/groups/trc/data/Ilana/2P/data/later/temp_filter'
```

```python
fly_list=['208', '209', '210', '217', '218', '226', '227', '228', '233', '234', '239', '240', '241', '242', '249', '250']
step_size=10
files = os.listdir(later_dir)
STA=np.load(os.path.join(later_dir,'giant_STA.npy'))
```

```python
np.shape(STA)
```

```python
%%time
total_giant_STA={}
count = len(fly_list)
for fly in fly_list:
    file_name = f"{fly}_stepsize_{step_size}_STA.h5"
    STA_path = os.path.join(later_dir,file_name)
    if os.path.exists(STA_path):
        with h5py.File(STA_path, 'r') as hf:
                brain = hf['data'][:]
                dimsb = np.shape(brain)
                print(f"Brain shape is {dimsb}")
                total_giant_STA[fly]=brain
total_giant_STA['total']=STA
```

```python
len(total_giant_STA.keys())
```

```python
fixed = brainsss.load_fda_meanbrain()
masked=np.where(fixed.numpy()[None,...]>0.1, total_giant_STA['total'], np.nan)
```

```python
plt.figure(figsize=(10,10))
plt.imshow(np.nanmax(masked[100][:,:,:], axis=-1).T)
```

```python
dims = np.shape(masked)
```

```python
reshape=dims[1]*dims[2]*dims[3]
reshape
```

```python
flat_total=np.reshape(masked, (dims[0], reshape))
flat_mean = np.nanmean(flat_total, axis=-1)
np.shape(flat_mean)
```

```python
fig, ax = plt.subplots(figsize=(10,10))
ax.plot(flat_mean)
# ax.set_ylim(-0.5, 1);
```

```python
fixed = brainsss.load_fda_meanbrain()
flattened={}
for key in total_giant_STA:
    print(key)
    masked=np.where(fixed.numpy()[None,...]>0.1, total_giant_STA[key], np.nan)
    nt, nx, ny, nz = np.shape(masked)
    reshape=nx*ny*nz
    flat_total=np.reshape(masked, (nt, reshape))
    flat_mean = np.nanmean(flat_total, axis=-1)
    flattened[key]=flat_mean
```

```python
fig, axs=plt.subplots(figsize=(20,10))
for key in flattened:
#     print(key)
    if key!='total':
        axs.plot(flattened[key]*10000, color='gray', alpha=0.3)
    else:
        axs.plot(flattened[key]*10000, color='k', lw=4)
axs.axvline(160,color='k',linestyle='--',lw=2)
axs.axvline(60,color='k',linestyle='--',lw=2)
```

```python
%%time
STA=STA ##### CHANGE THIS
explosions = []
for tp in range(np.shape(STA)[0]):
    input_canvas = np.ones((500,500,3)) #+.5 #.5 for diverging
    data_to_plot = STA[tp][:,:,::-1]
    vmax = 0.0005 #this was 0.5 for STA <------------
    explosion_map = brainsss.place_roi_groups_on_canvas(explosion_rois,
                                                        roi_masks,
                                                        roi_contours,
                                                        data_to_plot,
                                                        input_canvas,
                                                        vmax=vmax,
                                                        cmap='seismic', diverging=True)#'hot')
    explosions.append(explosion_map)
```
